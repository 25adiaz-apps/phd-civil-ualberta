The Breast Cancer Wisconsin (Diagnostic) dataset is a well-known and straightforward binary classification dataset in the field of machine learning. It is frequently utilized to showcase the performance of various algorithms. This dataset is included within the sklearn package and comprises a total of 569 samples, with measurements of 30 distinct features pertaining to breast cancer cell nuclei. The primary objective is to predict whether a tumor is malignant or benign. A comprehensive guide to this dataset can be accessed at the following link: https://scikit-learn.org/stable/datasets/toy_dataset.html#breast-cancer-dataset

For this assignment, you are required to apply clustering algorithms and Principal Component Analysis (PCA) to the dataset and address the subsequent questions. The following code snippet loads the dataset, divide it into training and testing sets and normalize the features for subsequent analysis. Please do not change the following code. **Please note that for this assignment, you are not required to create a validation set**

```python
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

# Load the breast cancer dataset
data = load_breast_cancer()

# Split the dataset into training and testing sets
X_train, X_test, y_train, y_test = train_test_split(data.data, data.target, test_size=0.5, random_state=20)

sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

```python
n_features = X_train.shape[1]
print("Number of features:", n_features)
print("Number of training samples:", X_train.shape[0])
print("Number of testing samples:", X_test.shape[0])
```

# Clustering Algorithms [10 Marks]

1.1 In this question, you will apply k-means algorithm to the **test set** directly. Report the **accuracy** of the algorithm by comparing the results to the known labels (use accuracy_score function in sklearn). **Note**: You are allowed to use sklearn package for algorithm implementation in this question. As the k-means algorithm is an unsupervised learning method, it does not utilize labels from the data. Consequently, the categories might be flipped (0 becomes 1, and 1 becomes 0). To address this issue, you can calculate both accuracy_score(y_test, kmeans.labels_) and accuracy_score(y_test, 1 - kmeans.labels_), then report the higher value. [4 Marks]

```python
# TODO: Please provide your answers here. You can add more cells for code or texts if needed. 
# Don't delete these two comments so that it would be easier for me to locate your answers.

from sklearn.cluster import KMeans
from sklearn.metrics import accuracy_score

# Building the KMeans model
kmeans = KMeans(n_clusters=2, random_state=20, n_init=100, max_iter=1000)
test_labels = kmeans.fit_predict(X_test)

accuracy_test_1 = accuracy_score(y_test, 1-test_labels)
accuracy_test_2 = accuracy_score(y_test, test_labels)
print(f"Test Accuracy: {max(accuracy_test_1, accuracy_test_2):.4f}")
```

1.2 In this question, you will apply the gaussian mixture model (GMM) to the **test set** directly. Report the **accuracy** of the algorithm by comparing the results to the known labels (use accuracy_score function in sklearn). **Note**: You are allowed to use sklearn package for algorithm implementation in this question. As the GMM is an unsupervised learning method, it does not utilize labels from the data. Consequently, the categories might be flipped (0 becomes 1, and 1 becomes 0). To address this issue, you can calculate both accuracy_score(y_test, kmeans.labels_) and accuracy_score(y_test, 1 - kmeans.labels_), then report the higher value. [4 Marks]

```python
# TODO: Please provide your answers here. You can add more cells for code or texts if needed. 
# Don't delete these two comments so that it would be easier for me to locate your answers.

from sklearn.mixture import GaussianMixture

gmm = GaussianMixture(n_components=2, covariance_type="full", n_init=10, random_state=20, max_iter=1000)
gmm.fit(X_test)

test_labels_gmm = gmm.predict(X_test)

accuracy_test_1 = accuracy_score(y_test, 1-test_labels_gmm)
accuracy_test_2 = accuracy_score(y_test, test_labels_gmm)
print(f"Test Accuracy: {max(accuracy_test_1, accuracy_test_2):.4f}")
```

# Principal Component Analysis [15 Marks]

2.1 Apply Principal Component Analysis (PCA) to the **training set** using **numpy package** only. Keep only one principal component. Report the percentage of variance explained by the first component. [5 Marks]

```python
# TODO: Please provide your answers here. You can add more cells for code or texts if needed. 
# Don't delete these two comments so that it would be easier for me to locate your answers.
import numpy as np

C_matrix = X_train.T @ X_train  / (X_train.shape[0] - 1)

U, S, Vt = np.linalg.svd(C_matrix)

variance_explained = S / np.sum(S)
percentage_explained = variance_explained * 100

Z_train_np = X_train @ Vt[0]

print(f"Variance explained by the first principal component: {np.max(S):.2f}")
print(f"Variance explained ratio by the first principal component: {np.max(percentage_explained):.2f}%")
```

2.2 Apply Principal Component Analysis (PCA) to the **training set** using **sklearn** package. Keep only one principal component. Report the percentage of variance explained by the first component. [4 Marks]

```python
# TODO: Please provide your answers here. You can add more cells for code or texts if needed. 
# Don't delete these two comments so that it would be easier for me to locate your answers.

from sklearn.decomposition import PCA

pca = PCA()
Z_train_sklearn = pca.fit_transform(X_train)

print(f"Variance explained by the first principal component: {pca.explained_variance_[0]:.2f}")
print(f"Variance explained ratio by the first principal component: {pca.explained_variance_ratio_[0] * 100:.2f}%")
```

2.2 Build a basic Support Vector Machine (SVM) model using both the original normalized features and features transformed by Principal Component Analysis (PCA), retaining 1, 5, 10, and 30 principal components. Compare the performance of these models on the test set. How does the number of principal components affect the performance of the model? Note: You are allowed to use the sklearn package for this question. [6 Marks]

```python
# TODO: Please provide your answers here. You can add more cells for code or texts if needed. 
# Don't delete these two comments so that it would be easier for me to locate your answers.

from sklearn.svm import SVC

kn = "linear"

# SVM for original data
svm_original = SVC(kernel=kn, C=1.0, gamma=0.01)
svm_original.fit(X_train, y_train)
y_pred_original = svm_original.predict(X_test)

accuracy_score(y_test, y_pred_original)
print(f"Accuracy of SVM for original data: {accuracy_score(y_test, y_pred_original):.4f}")

# SVM for PCA-reduced data

# For 1 component
Z_train_1 = pca.transform(X_train)[:, :1]
Z_test_1 = pca.transform(X_test)[:, :1]
svm_pca_1 = SVC(kernel=kn, C=1.0, gamma=0.01)
svm_pca_1.fit(Z_train_1, y_train)
y_pred_pca_1 = svm_pca_1.predict(Z_test_1)
print(f"Accuracy of SVM for 1 PCA component: {accuracy_score(y_test, y_pred_pca_1):.4f} ({sum(pca.explained_variance_ratio_[:1]) * 100:.2f}% variance explained)")

# For 5 components
Z_train_5 = pca.transform(X_train)[:, :5]
Z_test_5 = pca.transform(X_test)[:, :5]
svm_pca_5 = SVC(kernel=kn, C=1.0, gamma=0.01)
svm_pca_5.fit(Z_train_5, y_train)
y_pred_pca_5 = svm_pca_5.predict(Z_test_5)
print(f"Accuracy of SVM for 5 PCA components: {accuracy_score(y_test, y_pred_pca_5):.4f} ({sum(pca.explained_variance_ratio_[:5]) * 100:.2f}% variance explained)")

# For 10 components
Z_train_10 = pca.transform(X_train)[:, :10]
Z_test_10 = pca.transform(X_test)[:, :10]
svm_pca_10 = SVC(kernel=kn, C=1.0, gamma=0.01)
svm_pca_10.fit(Z_train_10, y_train)
y_pred_pca_10 = svm_pca_10.predict(Z_test_10)
print(f"Accuracy of SVM for 10 PCA components: {accuracy_score(y_test, y_pred_pca_10):.4f} ({sum(pca.explained_variance_ratio_[:10]) * 100:.2f}% variance explained)")

# For 30 components
Z_train_30 = pca.transform(X_train)[:, :30]
Z_test_30 = pca.transform(X_test)[:, :30]
svm_pca_30 = SVC(kernel=kn, C=1.0, gamma=0.01)
svm_pca_30.fit(Z_train_30, y_train)
y_pred_pca_30 = svm_pca_30.predict(Z_test_30)
print(f"Accuracy of SVM for 30 PCA components: {accuracy_score(y_test, y_pred_pca_30):.4f} ({sum(pca.explained_variance_ratio_[:30]) * 100:.2f}% variance explained)")

# For 16 components
Z_train_16 = pca.transform(X_train)[:, :16]
Z_test_16 = pca.transform(X_test)[:, :16]
svm_pca_16 = SVC(kernel=kn, C=1.0, gamma=0.01)
svm_pca_16.fit(Z_train_16, y_train)
y_pred_pca_16 = svm_pca_16.predict(Z_test_16)
print(f"Accuracy of SVM for 16 PCA components: {accuracy_score(y_test, y_pred_pca_16):.4f} ({sum(pca.explained_variance_ratio_[:16]) * 100:.2f}% variance explained)")
```

Taking the SVM model with the original data as a reference, it is shown that the number of PCA components selected has an impact on the accuracy of machine learning models. With 1 and 5 components the performance was lower (0.9193 and 0.9614). In this case, the number of components that gave the best performance was 10, which explain 95.25% of the variance of the data, even outperforming the SVM model with the original data. When all 30 PCA components were considered (k = n), 100% of the variance of the data is retained and the model is the same as the model with the original data, only with different axes, so it obtained the same performance. The selection of 16 components was added to compare the performance of the model when 99% of the variance of the data is taken into account, and this model also obtained the same performance as the model with 10 PCA components. The fact that the models with 10 and 16 components obtained an even better performance than the models that consider 100% of the variance of the data suggests that the discarded PCA components mainly represent variance due to noise.
