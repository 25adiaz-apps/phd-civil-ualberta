## 1. El puente

- Puente carretero de hormigón pretensado en Suiza, entre Utzenstorf y Koppigen. Cruzaba la autopista A1 Berna–Zúrich.
- Tres vanos: 14 + 30 + 14 m, más dos voladizos de 2.70 m.
- Tablero tipo cajón de dos celdas: 8.60 m de ancho y 1.10 m de canto.
- Dos pilas intermedias: **pila Utzenstorf** y **pila Koppigen**.
- Fue demolido a fines de 1998. Antes, el laboratorio EMPA hizo dos campañas de medición (proyecto europeo SIMCES):
  - **EMS** (*Environmental Monitoring System*): monitoreo continuo, cada hora, durante unos 10 meses.
  - **PDT** (*Progressive Damage Tests*): ensayos de vibración después de cada uno de 17 escenarios de daño.

## 3. Monitoreo continuo (EMS)

### 3.1 Qué se midió

- Cada hora, durante unos 10 meses: aceleraciones durante 655 s y variables ambientales antes y después.
- **5653 mediciones**, del 10.11.1997 al 10.09.1998. Hay huecos; el más largo, de 34 días, entre el 24.04 y el 28.05.1998.
- El daño de los PDT empezó el 11.08.1998, así que el último mes del EMS (674 horas) corresponde al puente dañado.
- Objetivo: ver cómo el ambiente, sobre todo la temperatura, cambia la vibración del puente.

### 3.2 Organización y formato

Cada medición horaria es un ZIP dentro de uno de los tres `Z24emsN.zip`:

```
Z24ems1.zip
├── 01C14.zip                 ← una medición horaria
│   ├── 01C1403.aaa           ← aceleración del canal 03
│   ├── 01C1405.aaa           ← ... uno por canal: 03, 05, 06, 07, 10, 12, 14, 16
│   ├── 01C14car.aaa          ← registro de paso de vehículos (no documentado)
│   ├── 01C14PRE.env          ← variables ambientales ANTES de las aceleraciones
│   └── 01C14POS.env          ← variables ambientales DESPUÉS de las aceleraciones
├── 01C15.zip
└── ...
```

**Nombre `WWDHH`:** `WW` = semana del monitoreo (desde 01), `D` = día de la semana (A = sábado, B = domingo, ..., G = viernes), `HH` = hora (00 a 23). Ejemplo: `01C14` = semana 1, lunes, 14:00.

**Archivos `.aaa` (aceleración)**, texto, un valor por línea:

| Línea | Contenido |
|---|---|
| 1 | Etiqueta |
| 2 | Número de muestras: 65 536 |
| 3 | Intervalo de muestreo: 0.01 s (100 Hz) |
| 4 en adelante | Las muestras |
| Final | Pie de texto con estadísticas (se ignora) |

Las 65 536 muestras se tomaron en 8 tramos de 8192 (82 s cada uno), que pueden tener pausas entre ellos. Las unidades no están indicadas.

**Archivos `.env` (ambientales)**, texto: una fila de encabezado con los nombres, 10 filas de lecturas (unos 14 s) y un pie de texto. Cada fila tiene 53 columnas.

**Archivo `car.aaa`:** no aparece en la documentación. Tiene el mismo formato que los `.aaa`, pero solo ceros y unos. Parece marcar el paso de vehículos. Está en el 72 % de las horas.

Son **49 sensores**, según la Tabla 11 de `Z24-EMS-readme.pdf`:

**Tabla 4.** Sensores ambientales del EMS.

| Grupo | Etiquetas | Cantidad | Qué mide |
|---|---|---|---|

### 3.5 A tener en cuenta

- El encabezado de los `.env` tiene 48 nombres para 53 columnas, y algunos nombres de la sección 3 están repetidos.
- Algunas columnas ambientales no tienen ningún dato válido (valores fijos como 3276.7 o −244.3), y el viento marca cero todo el tiempo.
- El acelerómetro 10 queda fijo en un valor constante en el 60 % de las horas (casi todo diciembre a abril).
- Cuando pasan vehículos pesados, varios acelerómetros se saturan (la señal se corta en un valor fijo).

## 4. Ensayos de daño progresivo (PDT)

### 4.1 Qué se midió

- Se aplicaron **17 escenarios** de daño, uno después de otro (el daño se acumula).
- Después de cada escenario se hicieron dos ensayos:
  - **AVT** (*Ambient Vibration Test*): vibración ambiental, sin excitación controlada.
  - **FVT** (*Forced Vibration Test*): vibración forzada con dos excitadores (*shakers*), midiendo también la fuerza aplicada.
- Cada ensayo se hizo en **9 setups**: como no había acelerómetros para cubrir todo el puente a la vez, se movieron por tramos.

| Escenario | Descripción |
|---|---|
| 01 | Primera medición de referencia (puente sano) |
| 02 | Segunda referencia, con el sistema para bajar la pila ya instalado |
| 03 | Descenso de la pila Koppigen, 20 mm |
| 04 | Descenso de la pila, 40 mm |
| 05 | Descenso de la pila, 80 mm |
| 06 | Descenso de la pila, 95 mm |
| 07 | Pila levantada de nuevo; inclinación de la fundación |

| 12 | Falla de las rótulas de hormigón en los estribos |

### 4.2 Organización y formato

```
pdt_01-08.zip
├── 01/
│   ├── avt/
│   │   ├── 01setup01.mat ... 01setup09.mat   ← 9 setups del ensayo ambiental
│   │   └── 168'349_701A.DOC                  ← informe del ensayo (Word)
│   └── fvt/
│       ├── 01setup01.mat ... 01setup09.mat   ← 9 setups del ensayo forzado
│       └── 168'349_701F.DOC
├── 02/ ... 08/
pdt_09_17.zip
└── 09/ ... 17/                               ← misma estructura
```

**Archivos `.mat`** (MATLAB), uno por setup, con dos variables:
- `data`: matriz de muestras × canales. 65 536 muestras (655 s a 100 Hz); en los escenarios 01 y 02, 65 530.
- `labelshulp`: el nombre de cada canal (columna), por ejemplo `202T`.

En total hay **306 archivos `.mat`** (17 escenarios × 2 ensayos × 9 setups).

**Excepciones:**
- El escenario 08 viene dos veces: en carpetas y en ZIP anidados (`08/avt.zip`, `08/fvt.zip`). Los archivos son idénticos.
- `06/avt`, `17/avt` y `17/FVT` traen además los mismos datos como `.aaa` sueltos (un archivo por canal).
- Los `.DOC` describen las convenciones de los sensores, pero la mayoría son una copia del escenario 01 y no describen el escenario que les corresponde.

### 4.3 Sensores y malla de medición

**Tablero** (Figura 4): tres líneas de puntos a lo largo del puente, con 45 posiciones cada una (de −1 a 43).

| Línea | Puntos | Ubicación | Direcciones medidas |
|---|---|---|---|
| 1 | 99 – 143 | Lado sur (Berna) | Vertical |
| 2 | 199 – 243 | Eje del tablero | Vertical y transversal; longitudinal en algunos puntos |
| 3 | 299 – 343 | Lado norte (Zúrich) | Vertical |

**Pilas** (Figura 5): 8 puntos en cada pila (4 alturas × 2 lados), con 3 direcciones cada uno.
- Pila Utzenstorf: puntos 411 – 442.
- Pila Koppigen: puntos 511 – 542.

**Referencias:** R1, R2 y R3, en la línea 1. No se mueven entre setups y sirven para unir los 9 setups. Son 5 canales: `R1V`, `R2L`, `R2T`, `R2V`, `R3V`.

**Fuerza (solo FVT):** `DP1V` y `DP2V`, la fuerza de los dos excitadores. DP1 está en el punto de R2 y DP2 en el de R1.

**En total:** 135 puntos en el tablero + 16 en las pilas = **151 puntos**, unos 290 canales por ensayo.

**Nombre de canal** = número de punto + dirección: `L` longitudinal, `T` transversal, `V` vertical. Ejemplo: `202T` = punto 202, dirección transversal.
- En el tablero: L hacia Koppigen, V hacia arriba, T hacia Berna.
- En las pilas los sensores están girados: **L es vertical** y V es horizontal.

### 4.4 A tener en cuenta

- Algunos setups tienen menos canales por errores de cableado (por ejemplo, el setup 7 del escenario 01 tiene 31). Para comparar escenarios hay que buscar los canales por su nombre en `labelshulp`, no por número de columna.
- En el FVT del escenario 02, la fuerza `DP2` debe multiplicarse por 6.25 (según su `.DOC`).
