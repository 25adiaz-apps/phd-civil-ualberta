## **1. Como verificar que dos variables son normalmente distribuidas por separado y en conjunto:**

Se revisa en dos pasos, porque una cosa no garantiza la otra. Si los datos son normales en conjunto, cada variable es normal por separado, pero no al revés.

### 1.1. Cada variable por separado (lo que se vio en clase con una variable)
- **Histograma:** debe tener forma de campana, simétrica y sin dos jorobas.
- **Gráfico Q-Q:** los puntos deben caer sobre la recta.
- **Prueba de Shapiro-Wilk:** un valor de $p$ alto, por ejemplo mayor que 0.05, significa que no hay evidencia en contra de la normalidad. No prueba que los datos sean normales.

### 1.2. Las dos en conjunto
- **Diagrama de dispersión:** la nube debe tener forma de elipse, sin curvaturas, sin grupos separados y sin forma de abanico.
- **Gráfico Q-Q de la distancia al cuadrado:** si los datos son normales en conjunto, la distancia de Mahalanobis al cuadrado de cada medición sana sigue una chi-cuadrado con 2 grados de libertad. Se ordenan esas distancias y se comparan con los cuantiles teóricos; deben caer sobre la recta.
- **Cobertura:** cerca del 95 % de los datos sanos debe quedar dentro de la elipse de 2.45.
- **Prueba formal:** la de Henze-Zirkler, disponible en la librería pingouin.
